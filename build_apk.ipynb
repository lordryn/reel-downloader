# 🚀 Build Reel Downloader Android APK with Buildozer

This notebook compiles the **Reel Downloader** KivyMD application into a native Android `.apk` package.

### Instructions:
1. Run **Step 1** to install build tools and Buildozer.
2. Run **Step 2** to upload `reel_downloader_src.zip` (located in your local `reel_downloader/` directory).
3. Run **Step 3** to build the debug APK.
4. Run **Step 4** to automatically download the compiled `.apk` file to your computer.

### Step 2: Upload and Unpack Source Code

```python
import os
from google.colab import files

if not os.path.exists('main.py'):
    print("Upload 'reel_downloader_src.zip' from your PC:")
    uploaded = files.upload()
    for fname in uploaded.keys():
        if fname.endswith('.zip'):
            !unzip -o "$fname"
else:
    print("Source files already present in working directory!")

!ls -la
```

### Step 3: Run Buildozer to Compile APK (~8-12 minutes on fresh build)

```python
# Accept Android SDK licenses and compile debug APK
!buildozer -v android debug
```

### Step 4: Download the Compiled APK

```python
import glob
from google.colab import files

apks = glob.glob('bin/*.apk')
if apks:
    print(f"\n Success! Downloading {apks[0]} to your local computer...")
    files.download(apks[0])
else:
    print("No APK found in bin/. Please check the Buildozer output in Step 3 for any errors.")
```
